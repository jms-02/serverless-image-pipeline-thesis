```python
# 셀 1 — 라이브러리
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import matplotlib.patches as mpatches
from scipy import stats
import numpy as np
# 한글 폰트 설정 (윈도우)
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False
```

```python
df1 = pd.read_csv('../data/exp1_size.csv')

# 기술통계
print(df1.groupby('size_kb')[['pillow_ms','rekog_ms','total_ms','e2e_ms']]
      .agg(['mean','std']).round(1))

# 선형 회귀
slope, intercept, r, p, se = stats.linregress(
    df1['size_kb'], df1['total_ms'])
print(f"회귀식: y = {slope:.4f}x + {intercept:.1f}")
print(f"R²={r**2:.4f}, p={p:.2e}")

# 그래프
fig, ax = plt.subplots(figsize=(8,5))
means = df1.groupby('size_kb')[['pillow_ms','rekog_ms']].mean()
x = np.arange(len(means))
ax.bar(x, means['pillow_ms'], label='Pillow', color='#378ADD', width=0.5)
ax.bar(x, means['rekog_ms'],  label='Rekognition', color='#BA7517',
       bottom=means['pillow_ms'], width=0.5)
ax.set_xticks(x)
ax.set_xticklabels([f'{kb}KB' for kb in means.index])
ax.set_xlabel('파일 크기'); ax.set_ylabel('처리 시간 (ms)')
ax.legend(); plt.tight_layout()
plt.savefig('../figures/exp1_analyze.png', dpi=300)
```

```python
df2 = pd.read_csv('../data/exp2_concurrency.csv')

# 기술통계
print(df2.groupby('concurrency')['total_ms']
      .agg(['mean','std','min','max']).round(1))

# 박스플롯
fig, ax = plt.subplots(figsize=(7,5))
groups = [df2[df2['concurrency']==c]['total_ms'].values
          for c in sorted(df2['concurrency'].unique())]
ax.boxplot(groups, tick_labels=[f'{c}장' for c in sorted(df2['concurrency'].unique())])
ax.set_xlabel('동시 업로드 수'); ax.set_ylabel('전체 완료 시간 (ms)')
plt.tight_layout()
plt.savefig('../figures/exp2_analyze.png', dpi=300)

# avg_per_file 분석 추가
print("\n=== 파일 1개당 평균 처리 시간 ===")
print(df2.groupby('concurrency')['avg_per_file']
      .agg(['mean','std']).round(1))

# 두 그래프를 나란히
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# 왼쪽: 배치 전체 시간
groups_total = [df2[df2['concurrency']==c]['total_ms'].values
                for c in sorted(df2['concurrency'].unique())]
ax1.boxplot(groups_total, tick_labels=['1장','5장','10장'])
ax1.set_title('배치 전체 완료 시간')
ax1.set_ylabel('시간 (ms)')

# 오른쪽: 파일 1개당 평균 시간 ← 오토스케일링 효과가 보이는 그래프
groups_avg = [df2[df2['concurrency']==c]['avg_per_file'].values
              for c in sorted(df2['concurrency'].unique())]
ax2.boxplot(groups_avg, tick_labels=['1장','5장','10장'])
ax2.set_title('파일 1개당 평균 처리 시간')
ax2.set_ylabel('시간 (ms)')

plt.tight_layout()
plt.savefig('../figures/exp2_analyze2.png', dpi=300)
plt.show()
```

```python
df3 = pd.read_csv('../data/exp3_coldstart.csv')

print("=== 기술통계 ===")
print(df3.groupby('state')['e2e_ms']
      .agg(['mean','std','min','max','count']).round(1))

warm = df3[df3['state'] == 'warm']['e2e_ms']
cold = df3[df3['state'] == 'cold']['e2e_ms']

t, p = stats.ttest_ind(cold, warm, equal_var=False)  # Welch's t-test
diff = cold.mean() - warm.mean()
ratio = (diff / warm.mean()) * 100
cohen_d = diff / warm.std()

print(f"\n=== 콜드 vs 웜 비교 ===")
print(f"웜  스타트 평균: {warm.mean():.1f}ms  표준편차: {warm.std():.1f}ms")
print(f"콜드 스타트 평균: {cold.mean():.1f}ms  표준편차: {cold.std():.1f}ms")
print(f"추가 지연:       {diff:.1f}ms ({ratio:.1f}%)")
print(f"t={t:.3f}, p={p:.2e}")
print(f"Cohen's d={cohen_d:.2f}")

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# 왼쪽: 박스플롯
axes[0].boxplot(
    [warm, cold],
    tick_labels=['웜 스타트', '콜드 스타트'],
    showmeans=True,
    meanprops={'marker': '^', 'markerfacecolor': 'green',
               'markeredgecolor': 'green', 'markersize': 8}
)
axes[0].set_ylabel('E2E 지연 시간 (ms)')
axes[0].set_title('웜 스타트 vs 콜드 스타트 분포')

# 오른쪽: 시도 순서별 꺾은선 (시간 흐름에 따른 패턴)
warm_vals = df3[df3['state']=='warm']['e2e_ms'].values
cold_vals = df3[df3['state']=='cold']['e2e_ms'].values

axes[1].plot(range(1, len(warm_vals)+1), warm_vals,
             'o-', color='#378ADD', label='웜 스타트', linewidth=1.5, markersize=5)
axes[1].plot(range(1, len(cold_vals)+1), cold_vals,
             's-', color='#D85A30', label='콜드 스타트', linewidth=1.5, markersize=5)
axes[1].axhline(warm.mean(), color='#378ADD', linestyle='--',
                alpha=0.6, label=f'웜 평균 {warm.mean():.0f}ms')
axes[1].axhline(cold.mean(), color='#D85A30', linestyle='--',
                alpha=0.6, label=f'콜드 평균 {cold.mean():.0f}ms')
axes[1].set_xlabel('시도 번호')
axes[1].set_ylabel('E2E 지연 시간 (ms)')
axes[1].set_title('시도 순서별 E2E 지연 시간')
axes[1].legend(fontsize=9)
axes[1].set_ylim(0, max(cold_vals) * 1.15)

plt.tight_layout()
plt.savefig('../figures/exp3_coldstart.png', dpi=300)
plt.show()

# 콜드 스타트 추가 지연 시각화
fig, ax = plt.subplots(figsize=(7, 5))

bar_warm = ax.bar(['웜 스타트', '콜드 스타트'],
                  [warm.mean(), warm.mean()],
                  color='#378ADD', label='기본 처리 시간', width=0.4)
bar_cold = ax.bar(['콜드 스타트'],
                  [diff],
                  bottom=[warm.mean()],
                  color='#D85A30', label=f'콜드 스타트 추가 지연 (+{diff:.0f}ms)',
                  width=0.4)

ax.errorbar(['웜 스타트'], [warm.mean()],
            yerr=[warm.std()], fmt='none', color='black', capsize=5)
ax.errorbar(['콜드 스타트'], [cold.mean()],
            yerr=[cold.std()], fmt='none', color='black', capsize=5)

ax.set_ylabel('E2E 지연 시간 (ms)')
ax.set_title('콜드 스타트 추가 지연 시간 구성')
ax.legend()
plt.tight_layout()
plt.savefig('../figures/exp3_coldstart_bar.png', dpi=300)
plt.show()
```

```python
df4 = pd.read_csv('../data/exp4_dlq.csv')

# 기본 통계
print("\n=== 처리 시간 통계 ===")
print(df4[['elapsed_ms']].describe().round(1))

print("\n=== DLQ 처리 결과 ===")
error_count = (df4['status'] == 'error').sum()
dlq_success = df4['dlq_increased'].sum()

print(f"에러 발생 수: {error_count}")
print(f"DLQ 도달 수: {dlq_success}")

print(f"에러율(%): {(df4['status'] == 'error').mean() * 100:.1f}%")
print(f"DLQ 성공률(%): {df4['dlq_increased'].mean() * 100:.1f}%")
```

```python
df5 = pd.read_csv('../data/exp5_rekognition.csv')

on  = df5[df5['rekog_enabled']==True]['total_ms']
off = df5[df5['rekog_enabled']==False]['total_ms']

on_mean, on_std   = on.mean(), on.std()
off_mean, off_std = off.mean(), off.std()

diff = on_mean - off_mean
ratio = (diff / on_mean) * 100

print(f"ON  평균: {on.mean():.1f}ms  표준편차: {on.std():.1f}ms")
print(f"OFF 평균: {off.mean():.1f}ms  표준편차: {off.std():.1f}ms")
print(f"추가 지연: {diff:.1f}ms ({ratio:.1f}%)")

t, p = stats.ttest_ind(on, off, equal_var=False)
print(f"t={t:.3f}, p={p:.2e}")

# 효과 크기 (Cohen's d)
cohen_d = diff / off_std
print(f"Cohen's d: {cohen_d:.2f}")

# 박스플롯
fig, ax = plt.subplots(figsize=(6,5))
ax.boxplot(
    [on, off],
    tick_labels=['Rekognition ON', 'Rekognition OFF'],
    showmeans=True
)
ax.set_ylabel('전체 처리 시간 (ms)')

plt.tight_layout()
plt.savefig('../figures/exp5_analyze.png', dpi=300)

plt.show()
```
