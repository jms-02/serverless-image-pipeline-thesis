```python
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
import numpy as np

df1 = pd.read_csv('analysis/data/exp1_size.csv')

# 기술통계
print(df1.groupby('size_kb')[['pillow_ms','rekog_ms','total_ms','e2e_ms']]
      .agg(['mean','std']).round(1))

# 선형 회귀
slope, intercept, r, p, se = stats.linregress(
    df1['size_kb'], df1['total_ms'])
print(f"회귀식: y = {slope:.4f}x + {intercept:.1f}")
print(f"R²={r**2:.4f}, p={p:.6f}")

# 그래프
fig, ax = plt.subplots(figsize=(8,5))
means = df1.groupby('size_kb')[['pillow_ms','rekog_ms']].mean()
x = np.arange(len(means))
ax.bar(x, means['pillow_ms'], label='Pillow', color='#378ADD', width=0.5)
ax.bar(x, means['rekog_ms'],  label='Rekognition', color='#BA7517',
       bottom=means['pillow_ms'], width=0.5)
ax.set_xticks(x)
ax.set_xticklabels([f'{kb}KB' for kb in means.index])
ax.set_xlabel('파일 크기'); ax.set_ylabel('처리 시간 (ms)')
ax.legend(); plt.tight_layout()
plt.savefig('analysis/figures/fig4_1.png', dpi=300)
```
